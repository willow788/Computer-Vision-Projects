```python
from termcolor import colored

#creating a comparator function
def comparator(learner, instructor):
  if len(learner)!= len(instructor):
    raise AssertionError("layer counting mismatch!")
  for a, b in zip(learner, instructor):
    if tuple(a) != tuple(b):
      raise AssertionError(colored('Test failed', attrs= ['bold']))
  print(colored('All tests passed', "green"))


#creating a summary function
def summary(model):
  results = []
  for layer in model.layers:
    output_shape = getattr(layer.output, 'shape', None)
    params = layer.count_params() if hasattr(layer, 'count_params') else 0
    results.append(layer.__class__.__name__ + ' ' + str(output_shape) + ' ' + str(params))
  return results


```
