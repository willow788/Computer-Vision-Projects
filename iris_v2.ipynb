```python
import kagglehub
path = kagglehub.dataset_download("arshid/iris-flower-dataset")
print('dataset downloaded')
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import cv2
from PIL import Image
import zipfile
import os
import tensorflow as tf
import keras
from keras import layers
from keras.models import load_model
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img
from keras.models import Sequential
from keras.layers import Conv2D, MaxPooling2D
from keras.layers import Activation, Dropout, Flatten, Dense
from keras.optimizers import Adam
print('all imports are successful')
```

```python
#printing where the file is downloaded
print(f"Dataset downloaded to: {path}")

#opening the csv file
csv_file = [f for f in os.listdir(path) if f.endswith('.csv')][0]
df = pd.read_csv(os.path.join(path, csv_file))
print(df.head())
print(df.info())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# 1. Separate features (X) and target label (y)
X = df.drop(columns=['species'])
y = df['species']

# 2. Encode the target labels (species strings -> integers)
encoder = LabelEncoder()
y_encoded = encoder.fit_transform(y)

# 3. Split the dataset into training and testing sets (e.g., 80% train, 20% test)
X_train, X_test, y_train, y_test = train_test_split(
    X, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded
)

print(f"Training set shape: {X_train.shape}")
print(f"Testing set shape: {X_test.shape}")
print(f"Unique classes: {encoder.classes_}")
```

```python
import keras
from keras.models import Sequential
from keras.layers import Dense, Input
from keras.optimizers import Adam

# 1. Define the Neural Network Architecture (with explicit Input layer for Keras 3 compatibility)
model = Sequential([
    Input(shape=(4,)),
    Dense(16, activation='relu'),
    Dense(12, activation='relu'),
    Dense(3, activation='softmax')
])

# 2. Compile the model
model.compile(
    optimizer=Adam(learning_rate=0.01),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# 3. Train the model
history = model.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=50,
    batch_size=8,
    verbose=1
)

# 4. Evaluate the model
loss, nn_accuracy = model.evaluate(X_test, y_test, verbose=0)
print(f"\nTest Accuracy: {nn_accuracy * 100:.2f}%")
```

```python
#training a logistic regression model on in
from sklearn.linear_model import LogisticRegression

#defining the model
model = LogisticRegression(max_iter=1000)

#fitting the model
model.fit(X_train, y_train)

#now training the model -- we will be fitting the training data to the model so it can learn from the parameters
model.predict(X_test)

#now check acc
lr_accuracy = model.score(X_test, y_test)
print(f"Accuracy: {lr_accuracy * 100:.2f}%")

if nn_accuracy > lr_accuracy:
  print('nn is better ')
else:
  print('lr is better')




```
